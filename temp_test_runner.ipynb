# GPU Test Runner

Run the repository test suite from the repository root and capture the results.

```python
import subprocess
import pathlib
import sys

root = pathlib.Path('.').resolve()
print(f'Running tests from: {root}')

result = subprocess.run([sys.executable, '-m', 'pytest', '-q'], cwd=root, capture_output=True, text=True)

print('=== STDOUT ===')
print(result.stdout)
print('=== STDERR ===')
print(result.stderr)
print('=== RETURN CODE ===')
print(result.returncode)

with open(root / 'pytest_output.txt', 'w', encoding='utf-8') as f:
    f.write('=== STDOUT ===
')
    f.write(result.stdout)
    f.write('
=== STDERR ===
')
    f.write(result.stderr)
    f.write(f'
=== RETURN CODE ===
{result.returncode}
')
```
